2. car = {"brand": "Toyota", "model": "Corolla", "year": 2018}. Add color='White', update year to 2022 and print dictionary.

```python
car = {
    "brand": "Toyota",
    "model": "Corolla",
    "year": 2018
}

car["color"] = "White"
car["year"] = 2022

print(car)
```

3. person = {"name": "Ali", "age": 25, "city": "Karachi"}. Print keys and values using loop.

```python
person = {
    "name": "Ali",
    "age": 25,
    "city": "Karachi"
}
for key, value in person.items():
    print(f"{key}: {value}")
```

4. numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]. Create frequency dictionary.

```python
numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]
frequency = {}

for num in numbers:
    frequency[num] = frequency.get(num, 0) + 1

print(frequency)
```

5. Take sentence input and count word frequency.

```python
sentence = input("Enter a sentence: ")
words = sentence.split()
word_count = {}

for word in words:
    word_clean = word.lower().strip(".,!?")
    if word_clean:
        word_count[word_clean] = word_count.get(word_clean, 0) + 1

print(word_count)
```

6. marks = {"Math": 80, "English": 70, "Physics": 90}. Calculate total and average.

```python
marks = {"Math": 80, "English": 70, "Physics": 90}

total = sum(marks.values())
average = total / len(marks)

print(f"Total Marks: {total}")
print(f"Average Marks: {average}")
```

7. students = {"Ali": 85, "Sara": 92, "Ahmed": 78, "Zara": 95}. Find highest scoring student.

```python
students = {
    "Ali": 85,
    "Sara": 92,
    "Ahmed": 78,
    "Zara": 95
}

top_student = max(students, key=students.get)
print(f"The highest scoring student is {top_student} with {students[top_student]} marks.")
```

8. dict1 = {"a": 1, "b": 2}, dict2 = {"c": 3, "d": 4}. Merge and print result.

```python
dict1 = {"a": 1, "b": 2}
dict2 = {"c": 3, "d": 4}

merged_dict = dict1 | dict2
print(merged_dict)
```

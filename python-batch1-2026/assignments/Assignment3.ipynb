```python
def print_alternate_elements():
    user_input = input("Enter elements separated by spaces: ")
    my_list = user_input.split()
    
    print("Alternate elements (starting from the first):")
    # Slicing with a step of 2 [start:end:step]
    for i in range(0, len(my_list), 2):
        print(my_list[i], end=" ")
    print()

# Uncomment to test:
# print_alternate_elements()
```

1. Write a program that accepts a list from user. Your list should reverse the content of list and display it. Do not use reverse() method.

```python
def reverse_list_manual():
    user_input = input("Enter elements separated by spaces: ")
    my_list = user_input.split()
    
    # Reversing using slicing (step -1)
    reversed_list = my_list[::-1]
    
    print("Original list:", my_list)
    print("Reversed list:", reversed_list)

# Uncomment to test:
# reverse_list_manual()
```

3. Find and display the largest number of a list without using built-in function max(). Your program should ask the user to input values in list from keyboard.

```python
def find_largest_number():
    user_input = input("Enter numbers separated by spaces: ")
    # Convert input strings to integers
    my_list = [int(x) for x in user_input.split()]
    
    if not my_list:
        print("List is empty!")
        return

    # Assume the first element is the largest
    largest = my_list[0]
    for num in my_list:
        if num > largest:
            largest = num
            
    print("The largest number is:", largest)

# Uncomment to test:
# find_largest_number()
```

4. Write a program that rotates the element of a list so that the element at the first index moves to the second index, the element in the second index moves to the third index, etc., and the element in the last index moves to the first index.

```python
def rotate_list():
    user_input = input("Enter elements separated by spaces: ")
    my_list = user_input.split()
    
    if not my_list:
        print("List is empty!")
        return
        
    # Pop the last element and insert it at the front (index 0)
    last_element = my_list.pop()
    my_list.insert(0, last_element)
    
    print("Rotated list:", my_list)

# Uncomment to test:
# rotate_list()
```

5. Write a program that input a string and ask user to delete a given word from a string.  


```python
def delete_word_from_string():
    sentence = input("Enter a sentence: ")
    word_to_delete = input("Enter the word you want to delete: ")
    
    # Split sentence into words, filter out the target word, and join back
    words = sentence.split()
    filtered_words = [w for w in words if w != word_to_delete]
    
    result = " ".join(filtered_words)
    print("Updated string:", result)

# Uncomment to test:
# delete_word_from_string()
```

6. Write a program that reads a string from the user containing a date in the form mm/dd/yyyy. It should print the date in the form March 12, 2021.

```python
def convert_date_format():
    date_str = input("Enter date in form mm/dd/yyyy: ")
    
    # List of month names corresponding to their numbers
    months = [
        "", "January", "February", "March", "April", "May", "June", 
        "July", "August", "September", "October", "November", "December"
    ]
    
    # Split the input string by '/'
    parts = date_str.split('/')
    if len(parts) != 3:
        print("Invalid format! Please use mm/dd/yyyy.")
        return
        
    month_num = int(parts[0])
    day = parts[1]
    year = parts[2]
    
    if 1 <= month_num <= 12:
        month_name = months[month_num]
        print(f"{month_name} {day}, {year}")
    else:
        print("Invalid month number!")

# Uncomment to test:
# convert_date_format()
```

8. Find the sum of each row of matrix of size m x n. For example for the following matrix output will be like this : 
Sum of row 1 = 32 
Sum of row 2 = 31 
Sum of row 3 = 63

```python
def row_sums_of_matrix():
    # Example matrix (3 rows, 3 columns)
    matrix = [
        [10, 12, 10],
        [5,  15, 11],
        [20, 23, 20]
    ]
    
    for i in range(len(matrix)):
        row_sum = sum(matrix[i])
        print(f"Sum of row {i + 1} = {row_sum}")

# Uncomment to test:
# row_sums_of_matrix()
```

9. Write a program to add two matrices of size n x m. 


```python
def add_matrices():
    # Example matrices of size 2 x 2
    A = [
        [1, 2],
        [3, 4]
    ]
    B = [
        [5, 6],
        [7, 8]
    ]
    
    rows = len(A)
    cols = len(A[0])
    
    # Initialize result matrix with zeros
    result = [[0 for _ in range(cols)] for _ in range(rows)]
    
    for i in range(rows):
        for j in range(cols):
            result[i][j] = A[i][j] + B[i][j]
            
    print("Sum of matrices:")
    for row in result:
        print(row)

# Uncomment to test:
# add_matrices()
```

10. Write a program to multiply two matrices

```python
def multiply_matrices():
    # Matrix A (2 x 3)
    A = [
        [1, 2, 3],
        [4, 5, 6]
    ]
    # Matrix B (3 x 2)
    B = [
        [7, 8],
        [9, 1],
        [2, 3]
    ]
    
    result = [[sum(a * b for a, b in zip(A_row, B_col)) for B_col in zip(*B)] for A_row in A]
    
    print("Product of matrices:")
    for row in result:
        print(row)

# Uncomment to test:
# multiply_matrices()
```

3. Write a program in which use all the operators we can use in Python

```python


a = 10
b = 3

print("--- 1. Arithmetic Operators ---")
print("Addition (a + b):", a + b)
print("Subtraction (a - b):", a - b)
print("Multiplication (a * b):", a * b)
print("Division (a / b):", a / b)
print("Modulus / Remainder (a % b):", a % b) 

print("\n--- 2. Assignment Operators ---")
x = 5      
print("Initial x:", x)
x += 3      
print("After x += 3:", x)
x *= 2      
print("After x *= 2:", x)

print("\n--- 3. Comparison Operators ---")
print("Is a greater than b? (a > b):", a > b)
print("Is a less than b? (a < b):", a < b)
print("Is a equal to b? (a == b):", a == b)
print("Is a not equal to b? (a != b):", a != b)

print("\n--- 4. Logical Operators ---")
cond1 = True
cond2 = False

print("True and False (and):", cond1 and cond2) 
print("True or False (or):", cond1 or cond2)   
print("Not True (not):", not cond1)           

print("\n--- 5. Bitwise Operators ---")
print("Bitwise AND (5 & 3):", 5 & 3)
print("Bitwise OR (5 | 3):", 5 | 3)
```

4. Completes the following steps of small task:

```python
English = 74
Islamiat = 86
Maths = 97
Total Marks = 300
obtained_marks = English + Islamiat + Maths
percentage = (obtained_marks / Total Marks) * 100

print("Obtained Marks:", obtained_marks)
print("Percentage:", percentage, "%")
```

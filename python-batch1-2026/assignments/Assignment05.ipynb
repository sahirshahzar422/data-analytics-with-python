Create a list called numbers containing at least 8 integers. Using a for loop, print all numbers, print the sum of all numbers, and print only the even numbers.

```python
# Question 1: Working with Lists
numbers = [4, 11, 16, 23, 8, 42, 15, 9]

print("--- All Numbers ---")
for num in numbers:
    print(num, end=" ")
print()

total_sum = 0
for num in numbers:
    total_sum += num

print(f"Sum of all numbers: {total_sum}")


print("Even numbers:")
for num in numbers:
    if num % 2 == 0:
        print(num, end=" ")
print("\n")
```

Create a dictionary called student with name, age, course, and marks (a list of 3 marks). Print all keys/values, calculate the average marks, and display whether the student Passed (average > 50) or Failed.

```python
# Question 2: Dictionary Basics
student = {
    "name": "Sahir",
    "age": 14,
    "course": "Computer Science",
    "marks": [75, 82, 90]
}

print("--- Student Details ---")
for key, value in student.items():
    print(f"{key}: {value}")

# Calculate the average of the marks using a loop
total_marks = 0
for mark in student["marks"]:
    total_marks += mark

average_marks = total_marks / len(student["marks"])
print(f"Average Marks: {average_marks:.2f}")

# Display Pass/Fail status
if average_marks >= 50:
    print("Status: Passed\n")
else:
    print("Status: Failed\n")
```

Create a list called employees with at least 3 employee dictionaries containing name, department, and salary. Print all details, find the employee with the highest salary, and calculate total salary expense.

```python
# Question 3: List of Dictionaries
employees = [
    {"name": "Zain", "department": "IT", "salary": 55000},
    {"name": "Bilal", "department": "Finance", "salary": 70000},
    {"name": "Hamza", "department": "HR", "salary": 45000}
]

print("--- Employee Details ---")
total_salary_expense = 0
highest_salary = 0
top_employee = ""

for emp in employees:
    print(f"Name: {emp['name']} | Dept: {emp['department']} | Salary: ${emp['salary']}")
    total_salary_expense += emp['salary']
    
    if emp['salary'] > highest_salary:
        highest_salary = emp['salary']
        top_employee = emp['name']

print(f"\nEmployee with Highest Salary: {top_employee} (${highest_salary})")
print(f"Total Salary Expense: ${total_salary_expense}\n")
```

Ask the user to enter numbers, store them in a list, and stop when -1 is entered. Print the complete list, the largest number, and the smallest number.

```python
# Question 4: While Loop with User Input
user_numbers = []
print("--- Number Collector (Enter -1 to stop) ---")

while True:
    try:
        user_input = int(input("Enter a number: "))
        if user_input == -1:
            break
        user_numbers.append(user_input)
    except ValueError:
        print("Invalid input! Please enter a valid integer.")

print(f"Complete List: {user_numbers}")
if len(user_numbers) > 0:
    print(f"Largest Number: {max(user_numbers)}")
    print(f"Smallest Number: {min(user_numbers)}\n")
else:
    print("No numbers were entered.\n")
```

Ask the user to enter a sentence, convert it into a list of words, use a dictionary to count occurrences, and print the word frequency clearly.

```python
# Question 5: Word Frequency Counter
sentence = input("Enter a sentence: ")

words_list = sentence.split()
word_counts = {}

for word in words_list:
    word_counts[word] = word_counts.get(word, 0) + 1

print("--- Word Frequency ---")
for word, count in word_counts.items():
    print(f"{word}: {count}")
print()
```

Use nested loops to print a multiplication table from 1 to 5 with proper formatting.

```python
# Question 6: Nested Loops – Multiplication Table
print("--- Multiplication Table (1 to 5) ---")

for i in range(1, 6):
    for j in range(1, 6):
        print(f"{i * j:<4}", end="")
    print() # Newline after each row
```

Create a DataFrame having some missing values (NaN). Find the count of missing values in each column, fill numeric NaNs with the column mean, and drop rows that still have missing values.   

```python
import numpy as np
import pandas as pd

# Creating a DataFrame with missing values
nan_data = {
    "Name": ["Aarav", "Priya", "Rahul", "Ananya", None],
    "Age": [28, np.nan, 45, 23, 31],
    "Salary": [45000, 62000, np.nan, 39000, 72000]
}
df_nan = pd.DataFrame(nan_data)

# Count missing values per column
print("Missing Value Counts:\n", df_nan.isnull().sum())

# Fill numeric NaNs with column mean
df_nan["Age"] = df_nan["Age"].fillna(df_nan["Age"].mean())
df_nan["Salary"] = df_nan["Salary"].fillna(df_nan["Salary"].mean())

# Drop rows that still have missing values (e.g., in 'Name')
df_cleaned = df_nan.dropna()

print("\nCleaned DataFrame:\n", df_cleaned)
```

Sort the DataFrame by Salary in descending order, and then by Age in ascending order. Also print the top 3 highest-paid employees using nlargest().

```python
# Sort by Salary descending, then Age ascending
sorted_df = df.sort_values(by=["Salary", "Age"], ascending=[False, True])
print("Sorted DataFrame:\n", sorted_df)

# Top 3 highest-paid employees using nlargest
top_3 = df.nlargest(3, "Salary")
print("\nTop 3 Highest-Paid Employees:\n", top_3)
```
